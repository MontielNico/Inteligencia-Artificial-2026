## Búsqueda Informada

```python
from EntornoLaberinto import LaberintoEnv
from AgenteBusquedaInf import AgenteGreedy, AgenteAEstrella

maze = [
    ["#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#"],
    ["#", "S", " ", " ", " ", " ", " ", " ", "#", " ", " ", " ", " ", " ", "#"],
    ["#", " ", "#", "#", "#", "#", "#", " ", "#", " ", "#", "#", "#", " ", "#"],
    ["#", " ", " ", " ", " ", " ", "#", " ", "#", " ", "#", " ", " ", " ", "#"],
    ["#", "#", "#", " ", "#", " ", "#", " ", "#", " ", "#", " ", "#", "#", "#"],
    ["#", " ", " ", " ", "#", " ", " ", " ", "#", " ", " ", " ", "#", "G", "#"],
    ["#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#", "#"],
]

entorno = LaberintoEnv(maze)

agenteGreedy = AgenteGreedy(entorno)
agenteAEstrella = AgenteAEstrella(entorno)
```

### **Busqueda Greedy**

```python
solucion_greedy = agenteGreedy.buscar()
print("--- RESULTADOS GREEDY ---")
if solucion_greedy:
    camino = agenteGreedy.camino_solucion
    entorno.imprimir_camino(camino)
    print("Camino encontrado:", agenteGreedy.camino_solucion)
    print("Costo Camino Encontrado:", len(agenteGreedy.camino_solucion))
else:
    print("NO SE ENCONTRO SOLUCION. Mostrando el recorrido explorado:")
    entorno.imprimir_camino(agenteGreedy.decisiones)

print("Nodos expandidos:", agenteGreedy.nodos_expandidos)
print("Retrocesos:", agenteGreedy.cantidad_retrocesos)
print("Total de decisiones:", len(agenteGreedy.decisiones))
print("Decisiones:", agenteGreedy.decisiones)
```

### **Busqueda A Estrella**

```python
solucion_a_estrella = agenteAEstrella.buscar()
print("--- RESULTADOS A* ---")
if solucion_a_estrella:
    camino = agenteAEstrella.camino_solucion
    entorno.imprimir_camino(camino)
    print("Camino encontrado:", agenteAEstrella.camino_solucion)
    print("Costo Camino Encontrado:", len(agenteAEstrella.camino_solucion))
else:
    print("NO SE ENCONTRO SOLUCION. Mostrando el recorrido explorado:")
    entorno.imprimir_camino(agenteAEstrella.decisiones)

print("Nodos expandidos:", agenteAEstrella.nodos_expandidos)
print("Retrocesos:", agenteAEstrella.cantidad_retrocesos)
print("Total de decisiones:", len(agenteAEstrella.decisiones))
print("Decisiones:", agenteAEstrella.decisiones)
```

Dentro de una heuristica que sea admisible, es decir que sea estricticamente optimista, nunca sobreestimamos el costo real para alcanzar el destino dentro de un nodo determinado. Matematicamente podemos expresarlos como h(n) <= h*(n), donde h(n) es expresado como el costo estimado por la heuristica y h*(n) es el costo real minimo
Por eso aqui entra la distancia Manhattan que es admisible, ya que este mismo en la situación de un laberinto, calcula la cantidad de pasos necesarios para llegar a la meta asumiento que no existirian obstaculos, cumpliendose matematicamente lo necesario, siempre sera menor o igual sin sobreestimarlo jamas.

### **Comparación de A Estrella frente Greedy**

- **¿Expande mas Nodos?**: El algoritmo Greedy es un algoritmo que busca el mejor camino posible sin importarle como llego alli, se lanza directamente al nodo al parecer mas cercano al destino. En cambio el A* calculo su camino, su costo al cual esta haciendo, si es mas costoso o no. Estas reflexiones nos dan que el algoritmo A* es mas costoso en expandir nodos, pero mas eficaz, ya que evalua, hace una exploración exhaustiva para encontrar rutas seguras.
- **¿Encuentra mejor solución?**: Como Greedy no compara su costo del camino, puede ser engañado por un camino que parece cercano pero lo desvia de su camino, obligandolo a terminar, en cambio A*, es mas audaz, gracias a tener una heurística admisible combinada con el costo real, nunca aceptara una solución final hasta no comparar datos para encontrar mejores soluciones
