## Búsqueda Local


```python
from EntornoSudoku import EntornoSudoku
from AgenteBusquedaLocal import AgenteLocal

sudoku_4x4 = [[0, 1, 0, 4], [0, 0, 3, 0], [0, 2, 0, 0], [0, 0, 0, 0]]

entorno = EntornoSudoku(sudoku_4x4)
agente = AgenteLocal(entorno)

print("ESTADO INICIAL 4x4")
entorno.imprimir_tablero(sudoku_4x4)
```

## **SUDOKU 4X4**

### **Hill Climbing**

```python
print("\nRESULTADO HILL CLIMBING (1 Corrida)")
nodo_hc, iter_hc = agente.hill_climbing()
entorno.imprimir_tablero(nodo_hc.tablero)
print(f"H Óptimo: {nodo_hc.h}")
print(f"Iteraciones: {iter_hc}")
```

### **Simulated Annealing**

```python
print("\nRESULTADO SIMULATED ANNEALING (1 Corrida)")
nodo_sa, iter_sa = agente.simulated_annealing()
entorno.imprimir_tablero(nodo_sa.tablero)
print(f"H Óptimo: {nodo_sa.h}")
print(f"Iteraciones: {iter_sa}")
```

### **SUDOKU 9X9**

```python
sudoku_9x9 = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9],
]

entorno = EntornoSudoku(sudoku_9x9)
agente = AgenteLocal(entorno)

print("ESTADO INICIAL 9x9")
entorno.imprimir_tablero(sudoku_9x9)

print("\nRESULTADO HILL CLIMBING (1 Corrida)")
nodo_hc, iter_hc = agente.hill_climbing()
entorno.imprimir_tablero(nodo_hc.tablero)
print(f"H Óptimo: {nodo_hc.h}")
print(f"Iteraciones: {iter_hc}")

print("\nRESULTADO SIMULATED ANNEALING (1 Corrida)")
nodo_sa, iter_sa = agente.simulated_annealing(
    temp_inicial=100.0, alpha=0.9999, max_iteraciones=500000
)  # Mas intentos para explorar ya que el espacio 9x9 es mas propenso a no encontrarse
entorno.imprimir_tablero(nodo_sa.tablero)
print(f"H Óptimo: {nodo_sa.h}")
print(f"Iteraciones: {iter_sa}")
```

## **EXPERIMENTACIÓN Y COMPARACIONES**

Dado que la busqueda local, entra dentro del dominio de lo Estocastico, de las probabilidad una sola ejecución del mismo no seria valido frente al azar. Es por eso que se realizaron N = 30 corridas independientes para cada algoritmos del mismo
- Calidad de Solución = Frente a los dos algoritmos **Simulated Annealing** presenta una mejor calidad de solución frente al Hill Climbing, (llegando a un h = 0), mientras que el Hill Climbing es mas propenso a estancarse en minimo locales
- Costos (Pasos necesarios) = Existe una gran diferencia frente a estos dos algoritmos, dentro de Simulated Annealing, es posible encontrar la solución optima, pero conlleva a tener una alta demanda del CPU, al estar explorando por todo el campo de los valores posibles. Hill Climbing de su parte conlleva menos pasos necesarios, con apenas unos pares de intentos antes de estancarse, dependiendo de la mera suerte

#### Analisis de los Fallos de Hill Climbing
Al ser un algoritmo meramente determinista dentro de su criterio de aceptación, solo acepta casos donde los movimientos son mejoras, donde el h es menor a que sus respectivos vecinos. En un entorno de Sudoku, donde esta lleno de minimos locales, el algoritmos al no ver mejoria se estanca.

#### **Sudoku 9x9: Escalabilidad**
Al trasladar este algoritmo al espacio de busqueda de una tablero 9x9 de Sudoku, se encuentran varios datos. El espacio de busqueda ahora es inmenso, con dimensiones de $10^{41}$ combinaciones aumentando la profundidad y cantidad de valores, con muchos minimos locales.
- Encontrando que el algoritmo de Hill Climbing, dentro de estos escenarios especificos se vuelve obsoleto al estar infectado de minimos locales por doquier.
- Con Simulated Annealing se encuentra que debemos de ser mas receptimos con los errores, al tener un espacio de busqueda inmenso, teniendo que aumentar la temperatura inicial y la tasa de enfriamiento del mismo, llevandonos a la conclusión, donde mas datos a explorar mas costos del hardware es requerida
